```python
#saving the datasets
from datasets import load_dataset
import os, gc

PROJECT = '/content/drive/MyDrive/Colab Notebooks/FreshRetailNet Demand Forecasting'
INPUTS  = f'{PROJECT}/inputs'
os.makedirs(INPUTS, exist_ok=True)

dataset = load_dataset("Dingdong-Inc/FreshRetailNet-50K")

dataset['train'].to_pandas().to_parquet(f'{INPUTS}/freshretailnet_train.parquet')
gc.collect()

dataset['eval'].to_pandas().to_parquet(f'{INPUTS}/freshretailnet_test.parquet')
del dataset; gc.collect()

print('saved ->', os.listdir(INPUTS))
```
